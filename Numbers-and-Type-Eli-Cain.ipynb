```python
seconds = int(60)
minutes = int(60)
seconds_per_hour = seconds * minutes
print("There are", seconds_per_hour, "seconds in an hour.")
seconds_per_day = seconds_per_hour * 24
print("There are", seconds_per_day, "seconds in a day.")
result1 = seconds_per_day / seconds_per_hour
result2 = seconds_per_day // seconds_per_hour
print("This is the result using floating point division:", result1)
print("This is the result using integer division:", result2)
print("The result of the floating point division is equal to the result of the integer division:", result1 == result2)
```
